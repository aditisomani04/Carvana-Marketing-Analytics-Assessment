## Question 1:

```python
%%sql
    
with 
channel_category AS (
    select channel_id, TRIM(SUBSTR(channel, 1, INSTR(channel, '-') - 1)) AS category
    from ad_channels),
click_counts AS (
    select channel_id, count(*) as total_clicks
    from clicks 
    group by channel_id),
spent_totals AS (
    select channel_id, SUM(spend) as total_spend
    from spend
    group by channel_id)

SELECT 
    cc.category,
    SUM(s.total_spend) AS total_spend,
    SUM(c.total_clicks) AS total_clicks,
    ROUND(SUM(s.total_spend) * 1.00 / SUM(c.total_clicks),4) as cost_per_click
FROM channel_category cc
JOIN  click_counts  c ON c.channel_id = cc.channel_id
JOIN  spent_totals s ON s.channel_id = cc.channel_id
GROUP BY cc.category
ORDER BY cost_per_click ASC
```

## Question 2:

```python
%%sql

WITH first_touch AS (
    SELECT
        user_id,
        channel_id,
        ROW_NUMBER() OVER (PARTITION BY user_id ORDER BY click_datetime ASC) AS rn
    FROM clicks
),
channel_parts AS (
    -- category - partner _ campaign parsing; if there's no "_" the partner is the campaign
    SELECT
        channel_id,
        TRIM(SUBSTR(channel, 1, INSTR(channel, '-') - 1)) AS category,
        CASE WHEN INSTR(SUBSTR(channel, INSTR(channel,'-')+1), '_') > 0
             THEN TRIM(SUBSTR(SUBSTR(channel, INSTR(channel,'-')+1), 1, INSTR(SUBSTR(channel, INSTR(channel,'-')+1), '_') - 1))
             ELSE TRIM(SUBSTR(channel, INSTR(channel,'-')+1)) END AS partner,
        CASE WHEN INSTR(SUBSTR(channel, INSTR(channel,'-')+1), '_') > 0
             THEN TRIM(SUBSTR(SUBSTR(channel, INSTR(channel,'-')+1), INSTR(SUBSTR(channel, INSTR(channel,'-')+1), '_') + 1))
             ELSE TRIM(SUBSTR(channel, INSTR(channel,'-')+1)) END AS campaign
    FROM ad_channels
)
SELECT
    cp.category, cp.partner, cp.campaign,
    COUNT(*) AS num_sales
FROM sales s
JOIN first_touch ft ON ft.user_id = s.user_id AND ft.rn = 1
JOIN channel_parts cp ON cp.channel_id = ft.channel_id
GROUP BY cp.channel_id
ORDER BY num_sales DESC
LIMIT 3;
```

## Question 3:

```python
%%sql

WITH first_touch AS (
    SELECT user_id, channel_id,
        ROW_NUMBER() OVER (PARTITION BY user_id ORDER BY click_datetime ASC) AS rn
    FROM clicks
),
channel_parts AS (
    SELECT
        channel_id,
        TRIM(SUBSTR(channel, 1, INSTR(channel, '-') - 1)) AS category,
        CASE WHEN INSTR(SUBSTR(channel, INSTR(channel,'-')+1), '_') > 0
             THEN TRIM(SUBSTR(SUBSTR(channel, INSTR(channel,'-')+1), 1, INSTR(SUBSTR(channel, INSTR(channel,'-')+1), '_') - 1))
             ELSE TRIM(SUBSTR(channel, INSTR(channel,'-')+1)) END AS partner
    FROM ad_channels
),
spend_totals AS (
    SELECT cp.category, cp.partner, SUM(sp.spend) AS total_spend
    FROM spend sp JOIN channel_parts cp ON cp.channel_id = sp.channel_id
    GROUP BY cp.category, cp.partner
),
sales_by_partner AS (
    SELECT cp.category, cp.partner, COUNT(*) AS num_sales
    FROM sales s
    JOIN first_touch ft ON ft.user_id = s.user_id AND ft.rn = 1
    JOIN channel_parts cp ON cp.channel_id = ft.channel_id
    GROUP BY cp.category, cp.partner
)
SELECT
    st.category, st.partner,
    ROUND(st.total_spend, 2) AS total_spend,
    sbp.num_sales,
    ROUND(st.total_spend * 1.0 / sbp.num_sales, 2) AS cac
FROM spend_totals st
JOIN sales_by_partner sbp ON sbp.partner = st.partner AND sbp.category = st.category
ORDER BY cac ASC;
```

## Question 4:

```python
%%sql
    
WITH
-- Average APR per month, financed sales only (used for the APR modifier)
monthly_apr AS (
    SELECT STRFTIME('%Y-%m', sale_datetime) AS sale_month,
           AVG(apr) AS avg_apr
    FROM sales
    WHERE is_financed = 1
    GROUP BY sale_month
),

-- Profit per sale, following the APR modifier
sales_profit AS (
    SELECT
        s.user_id,
        s.sale_id,
        v.avg_margin *
        (1 + CASE WHEN s.is_financed = 1
                  THEN (s.apr - m.avg_apr) / m.avg_apr
                  ELSE -0.1 END)
        AS modified_value,
        v.bodystyle,
        s.has_trade_in,
        s.delivery_distance
    FROM sales s
    JOIN vehicles v
        ON LOWER(s.make) = LOWER(v.make) AND LOWER(s.model) = LOWER(v.model)
    LEFT JOIN monthly_apr m
        ON m.sale_month = STRFTIME('%Y-%m', s.sale_datetime)
),

profit_final AS (
    SELECT
        user_id,
        CASE
            WHEN bodystyle IN ('Sedan','Hatchback')
                THEN modified_value + 200 - delivery_distance/2.0
                     + CASE WHEN has_trade_in = 1 THEN 400 ELSE 0 END
            WHEN bodystyle IN ('Coupe','SUV')
                THEN modified_value - 0.8*delivery_distance
                     + CASE WHEN has_trade_in = 1 THEN 300 ELSE 0 END
            WHEN bodystyle = 'Truck'
                THEN modified_value - 200 - delivery_distance
                     + CASE WHEN has_trade_in = 1 THEN 200 ELSE 0 END
        END AS profit
    FROM sales_profit
),

-- First touch channel per user
first_touch AS (
    SELECT user_id, channel_id,
           ROW_NUMBER() OVER (PARTITION BY user_id ORDER BY click_datetime ASC) AS rn
    FROM clicks
),

-- Parse category out of channel name
channel_category AS (
    SELECT channel_id, TRIM(SUBSTR(channel, 1, INSTR(channel, '-') - 1)) AS category
    FROM ad_channels
),

spend_totals AS (
    SELECT cc.category, SUM(sp.spend) AS total_spend
    FROM spend sp
    JOIN channel_category cc ON cc.channel_id = sp.channel_id
    GROUP BY cc.category
)

SELECT
    cc.category,
    ROUND(SUM(pf.profit), 2) AS total_profit,
    st.total_spend,
    ROUND(SUM(pf.profit) * 1.0 / st.total_spend, 2) AS roi
FROM profit_final pf
JOIN first_touch ft ON ft.user_id = pf.user_id AND ft.rn = 1
JOIN channel_category cc ON cc.channel_id = ft.channel_id
JOIN spend_totals st ON st.category = cc.category
GROUP BY cc.category, st.total_spend
ORDER BY roi DESC;
```

```python
%%sql
    
WITH clicks_before_lock AS (
select  c.user_id, c.channel_id
from clicks c
JOIN locks l on l.user_id = c.user_id 
where c.click_datetime < l.lock_datetime),

channel_per_user AS (
select user_id, COUNT(distinct channel_id)  as num_channels
from clicks_before_lock
group by user_id)

select ROUND(AVG(num_channels),2) as avg_num_channels
from channel_per_user
```

```python
#Python

clicks = pd.read_csv("clicks.csv")
sales = pd.read_csv("sales.csv")

clicks["click_datetime"] = pd.to_datetime(clicks["click_datetime"])
sales["sale_datetime"] = pd.to_datetime(sales["sale_datetime"], format="%m/%d/%Y %H:%M")

merged = sales[["sale_id", "user_id", "sale_datetime"]].merge(
    clicks[["user_id", "click_datetime"]], on="user_id", how="left"
)
merged["in_window"] = (
    merged["click_datetime"].notna()
    & (merged["click_datetime"] >= merged["sale_datetime"] - pd.Timedelta(days=90))
    & (merged["click_datetime"] <= merged["sale_datetime"])
)

has_click = merged.groupby("sale_id")["in_window"].any()
sales = sales.merge(has_click.rename("has_recent_click"), on="sale_id")

total = len(sales)
no_click = (~sales["has_recent_click"]).sum()
pct = round(100 * no_click / total, 2)

print(f"{no_click} of {total} sales ({pct}%) have no click in the prior 90 days")
```

## Question 7:

```python
#Python 

clicks = pd.read_csv("clicks.csv")
sales = pd.read_csv("sales.csv")
spend = pd.read_csv("spend.csv")
channels = pd.read_csv("ad_channels.csv")
vehicles = pd.read_csv("vehicles.csv")
sales["make"] = sales["make"].str.strip().str.title()
vehicles["make"] = vehicles["make"].str.strip().str.title()

channels["category"] = channels["channel"].str.split("-").str[0].str.strip()

clicks["click_datetime"] = pd.to_datetime(clicks["click_datetime"])
sales["sale_datetime"] = pd.to_datetime(sales["sale_datetime"], format="%m/%d/%Y %H:%M")

# first touch
first_touch = (clicks.sort_values("click_datetime")
                      .groupby("user_id").first().reset_index())
first_touch = first_touch.merge(channels[["channel_id", "category"]], on="channel_id")

# profit per sale
sales["sale_month"] = sales["sale_datetime"].dt.to_period("M")
monthly_apr = sales.loc[sales.is_financed == 1].groupby("sale_month")["apr"].mean()

s = sales.merge(vehicles, on=["make", "model"])
s["apr_mod"] = s.apply(
    lambda r: (r.apr - monthly_apr[r.sale_month]) / monthly_apr[r.sale_month]
    if r.is_financed == 1 else -0.1, axis=1
)
s["mv"] = s["avg_margin"] * (1 + s["apr_mod"])

def profit(r):
    if r.bodystyle in ("Sedan", "Hatchback"):
        p = r.mv + 200 - r.delivery_distance / 2
        return p + 400 if r.has_trade_in else p
    if r.bodystyle in ("Coupe", "SUV"):
        p = r.mv - 0.8 * r.delivery_distance
        return p + 300 if r.has_trade_in else p
    if r.bodystyle == "Truck":
        p = r.mv - 200 - r.delivery_distance
        return p + 200 if r.has_trade_in else p

s["profit"] = s.apply(profit, axis=1)
s = s.merge(first_touch[["user_id", "category"]], on="user_id") 

spend["category"] = spend.merge(channels, on="channel_id")["category"]
clicks_cat = clicks.merge(channels, on="channel_id")

summary = (spend.groupby("category")["spend"].sum().to_frame()
    .join(clicks_cat.groupby("category").size().rename("clicks"))
    .join(s.groupby("category").size().rename("num_sales"))
    .join(s.groupby("category")["profit"].sum().rename("profit")))

summary["cpc"] = (summary["spend"] / summary["clicks"]).round(2)
summary["cac"] = (summary["spend"] / summary["num_sales"]).round(0)
summary["roi"] = (summary["profit"] / summary["spend"]).round(2)

print(summary[["spend", "cpc", "cac", "roi"]].sort_values("roi", ascending=False))
```
